## Insights


### 1 Correlación

```python
numerical_data = rd.select_dtypes(include=[np.number])
data_correlation = numerical_data.corr()

data_correlation_filtered = data_correlation.where(abs(data_correlation) > 0.4)

plt.figure(figsize=(10, 8))
sns.heatmap(data_correlation_filtered, annot=True, cmap='coolwarm', center=0)
sns.set(font_scale=1.0, style="whitegrid")
plt.title('Correlation Heatmap (|correlation| > 0.4)')
plt.show()
```

```python
### 2 customer age vs months on book
plt.figure(figsize=(8, 6))
sns.scatterplot(data=rd, x='customer_age', y='months_on_book', color ='blue', alpha=0.6)
plt.title('Customer Age vs Months on Book')
plt.xlabel('Customer Age')
plt.ylabel('Months on Book')
plt.show()
```

```python
## 3 Perfil general de los clientes  
plt.figure(figsize=(10, 6))
sns.histplot(data=rd, x='credit_limit', bins=30, kde=True, color='green')
plt.title('Distribution of Credit Limit')
plt.xlabel('Credit Limit')
plt.ylabel('Frequency')
plt.show()
```

```python
## 4 Distribution of Dependent Count
plt.figure(figsize=(10, 6))
sns.countplot(data=rd, x='dependent_count', palette='viridis', hue='dependent_count')
plt.title('Distribution of Dependent Count')
plt.xlabel('Dependent Count')
plt.ylabel('Frequency')
plt.show()
```

```python
#5 Contacts Count vs Attrition Flag
plt.figure(figsize=(8, 6))
sns.countplot(data=rd, x='contacts_count_12_mon', hue='attrition_flag', palette='Set1')
plt.title('Contacts Count vs Attrition Flag')
plt.xlabel('Contacts Count in Last 12 Months')
plt.ylabel('Number of Customers')
plt.show()
```

```python
#6 Dependent count per Income category
# Calcular el promedio por categoría
avg_dep = rd.groupby('income_category')['dependent_count'].mean().reset_index()

# Ordenar de mayor a menor
avg_dep = avg_dep.sort_values('dependent_count', ascending=False)

# Crear una paleta degradada (del mismo color, de más intenso a más claro)
palette = sns.color_palette("Blues", n_colors=len(avg_dep))

# Invertir la paleta para que el mayor tenga más color (más oscuro)
palette = palette[::-1]

# Graficar
plt.figure(figsize=(10, 6))
sns.barplot(
    data=avg_dep,
    x='income_category',
    y='dependent_count',
    palette=palette,
    errorbar=None
)

plt.title('Average Dependent Count per Income Category')
plt.xlabel('Income Category')
plt.ylabel('Average Dependent Count')
plt.xticks(rotation=45)
plt.show()
```

### Boxplots

```python
# Months on book vs Attrition Flag
plt.figure(figsize=(8, 6))
sns.violinplot(data=rd, x='attrition_flag', y='months_on_book', palette='pastel', inner='quartile')

# compute and plot means on top of the violinplot
means = rd.groupby('attrition_flag')['months_on_book'].mean().reset_index()
sns.scatterplot(data=means, x='attrition_flag', y='months_on_book', color='k', s=100, zorder=10, marker='D')

plt.title('Months on Book by Attrition Flag')
plt.xlabel('Attrition Flag')
plt.ylabel('Months on Book')
plt.show()
```

```python
# total count change q4 q1 vs attrition flag
plt.figure(figsize=(8, 6))
sns.violinplot(data=rd, x='attrition_flag', y='total_ct_chng_q4_q1', palette='Set2', inner='quartile')
plt.title('Total Count Change Q4 to Q1 by Attrition Flag')
plt.xlabel('Attrition Flag')
plt.ylabel('Total Count Change Q4 to Q1')
plt.show()
```
